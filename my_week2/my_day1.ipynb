## new trio conversation between me and my two sister
iam menna **

 Sara (GPT) → Menna (Gemma) → Nada (LLaMA) and repeat.

```python
import openai
import ollama
import random


# AI Models
sara_model = "gpt-4o-mini"     # GPT
menna_model = "gemma:2b"        # Gamma / Gemma
nada_model = "llama3.2:3b"      # LLaMA

# System Prompts
sara_system = (
    "You are Sara, the oldest sister. You are quiet, calm, and responsible. "
    "You always try to solve problems and help everyone in the house, even when your sisters fight."
)

menna_system = (
    "You are Menna, the middle sister who is often forgotten. You are independent, smart, and study AI. "
    "You go out a lot with your friends and act like you don’t care, but you secretly want recognition."
)

nada_system = (
    "You are Nada, the youngest spoiled sister—mama’s favorite. You refuse to do chores, "
    "constantly fight with Menna, and spend your time playing video games. You are dramatic and petty."
)

# Conversation history (full memory)
conversation_history = [
    {"role": "user", "content": "Girls, Mom said one of us has to clean the kitchen. Who’s doing it?"}
]
```

```python

def call_sara():
    response = openai.chat.completions.create(
        model=sara_model,
        messages=[{"role": "system", "content": sara_system}] + conversation_history
    )
    return response.choices[0].message.content

def call_menna():
    response = ollama.chat(
        model=menna_model,
        messages=[{"role": "system", "content": menna_system}] + conversation_history
    )
    return response['message']['content']

def call_nada():
    response = ollama.chat(
        model=nada_model,
        messages=[{"role": "system", "content": nada_system}] + conversation_history
    )
    return response['message']['content']
```

```python
turn_order = ["sara", "menna", "nada"]
last_current_speaker = None

for i in range(6):
    current_speaker = random.choice(turn_order)

    while current_speaker == last_current_speaker:  # re-roll if same as previous
        current_speaker = random.choice(turn_order)

    if current_speaker == "sara":
        reply = call_sara()
        print(f"\nSara: {reply}")

    elif current_speaker == "menna":
        reply = call_menna()
        print(f"\nMenna: {reply}")

    else:  # nada
        reply = call_nada()
        print(f"\nNada: {reply}")

    conversation_history.append({"role": "assistant", "content": reply})
    conversation_history.append({"role": "user", "content": reply})

    last_current_speaker = current_speaker
```
